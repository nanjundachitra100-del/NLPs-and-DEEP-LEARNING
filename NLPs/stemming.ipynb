```python
#stemming is the process of reducing a word into its root format.
from nltk.stem import PorterStemmer

stemmer= PorterStemmer()
words=["eating","eat","eaten","looked","looking","lookes"]
for word in words:
    print(word+"--->"+stemmer.stem(word))
```

```python
stemmer.stem("congradulations")

##this is the disadvantage of Porterstemmer .For somewods it wont give the proper words
```

```python
from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)

print(reg_stemmer.stem('eating'))
print(reg_stemmer.stem('eatable'))
```

```python
from nltk.stem import SnowballStemmer
##It will come out of the disadvantage faced by the PorterStemmer

SnowballStemmer=SnowballStemmer('english')

words=["eating","eat","eaten","had","having"]
for word in words:
    print(word+"--->"+SnowballStemmer.stem(word))
```
