```python
import re
#re is regular expression
def clean_text(text):
    text = str(text)
    #converting to lowercase
    text = text.lower()
    #remove the unwanted special cases
    text = re.sub(r'[^\w\s]', '',text)
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
df["clean_text"]=df["text"].apply(clean_text)
```

```python
df[["text", "clean_text", "sentiment"]].head(10)
```

```python
import nltk
from nltk.tokenize import word_tokenize

df["tokenize"]=df["clean_text"].apply(word_tokenize)

df[["clean_text", "tokenize", "sentiment"]].head(10)
```
