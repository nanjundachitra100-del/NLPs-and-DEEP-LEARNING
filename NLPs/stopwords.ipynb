```python
import nltk
nltk.download('stopwords')
```

```python
paragraph = """Technology is changing the way people learn and communicate. 
Students are using computers and smartphones to access educational resources, 
watch videos, and practice new skills. Learning new technologies can help 
students build confidence and prepare for exciting career opportunities."""

import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

stemmer = PorterStemmer()

sentence = nltk.sent_tokenize(paragraph)

for i in range(len(sentence)):

    # Word tokenization
    words = nltk.word_tokenize(sentence[i])

    # Remove stopwords and apply stemming
    words = [
        stemmer.stem(word)
        for word in words
        if word.lower() not in stop_words
    ]

    # Join words back into a sentence
    sentence[i] = ' '.join(words)

print(sentence)
```
